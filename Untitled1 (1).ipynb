# Level 4 – Task 8: Investment Duration

## Objective
To calculate the average investment duration of the participants.

```python
# Check the different investment duration categories

duration_counts = df['Duration'].value_counts()

print(duration_counts)
```

```python
# Convert duration categories into representative numerical values

duration_mapping = {
    'Less than 1 year': 0.5,
    '1-3 years': 2,
    '3-5 years': 4,
    'More than 5 years': 6
}

df['Duration_Numeric'] = df['Duration'].map(duration_mapping)

print(df[['Duration', 'Duration_Numeric']].head())
```

```python
# Calculate average investment duration

average_duration = df['Duration_Numeric'].mean()

print("Average Investment Duration:", round(average_duration, 2), "years")
```

```python
duration_counts.plot(kind='bar')

plt.title('Investment Duration of Participants')
plt.xlabel('Investment Duration')
plt.ylabel('Number of Participants')
plt.xticks(rotation=45)
plt.show()
```

## Task 8 Findings

The analysis shows that the most common investment duration is 3–5 years, selected by 19 participants, followed by 1–3 years, selected by 18 participants.

Only 2 participants reported an investment duration of less than 1 year, while 1 participant reported an investment duration of more than 5 years.

By assigning representative values to the duration ranges, the estimated average investment duration is approximately 2.98 years.

Overall, the participants generally prefer medium-term investment horizons, with most participants investing for between 1 and 5 years.

# Level 4 – Task 10: Correlation Analysis

## Objective
To explore potential correlations between factors such as age, investment duration, and expected returns.

```python
# Convert expected return ranges into representative numerical values

expect_mapping = {
    '10%-20%': 15,
    '20%-30%': 25,
    '30%-40%': 35
}

df['Expected_Return_Numeric'] = df['Expect'].map(expect_mapping)

df[['Expect', 'Expected_Return_Numeric']].head()
```

```python
# Select variables for correlation analysis

correlation_data = df[
    ['age', 'Duration_Numeric', 'Expected_Return_Numeric']
]

correlation_data.corr()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(7, 5))

sns.heatmap(
    correlation_data.corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Between Age, Investment Duration and Expected Returns')
plt.show()
```

```python
correlation_matrix = correlation_data.corr()

print(correlation_matrix.round(2).to_string())
```

## Task 10 Findings

The correlation analysis was performed to examine the relationships between age, investment duration, and expected returns.

The correlation between age and investment duration is 0.05, indicating a very weak positive relationship. This suggests that age has little linear relationship with the investment duration in this dataset.

The correlation between age and expected returns is -0.09, indicating a very weak negative relationship. Therefore, age does not appear to have a meaningful linear relationship with expected return expectations.

The correlation between investment duration and expected returns is 0.26, indicating a weak positive relationship. This suggests that participants with longer investment durations tend to have slightly higher expected returns, although the relationship is relatively weak.

Overall, the correlations indicate that there are no strong linear relationships among age, investment duration, and expected returns in this dataset.
